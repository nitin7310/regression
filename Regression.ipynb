Q1.What is Simple Linear Regression?

  -> Simple Linear Regression is a fundamental statistical technique used to model the relationship between two variables by fitting a straight line to the observed data. It helps predict one variable (the dependent variable) based on the value of another (the independent variable). The equation for a simple linear regression model is:

              y = mx + b


Q2.What are the key assumptions of Simple Linear Regression?

  -> Linearity – The relationship between the independent variable ( x )
     and the dependent variable ( y ) should be linear, meaning a straight-line relationship exists.

     Independence – Observations should be independent of each other. This means that the value of one data point does not influence another.

     Homoscedasticity – The variance of residuals (differences between observed and predicted values) should be constant across all levels of the independent variable.

     Normality of Residuals – Residuals (errors) should be normally distributed. This is particularly important for making reliable statistical inferences.

     No Multicollinearity – Since Simple Linear Regression involves only one independent variable, multicollinearity isn't typically an issue. However, in multiple regression, predictors should not be highly correlated.

Q18.What are the limitations of using R² as a sole measure of model performance?

  -> Limitations of R²
Does Not Indicate CausationA high 𝑅2 suggests a strong association but does not confirm causality between independent and dependent variables.

Ignores Model Complexity𝑅2doesn’t account for overfitting, where the model fits the training data too well but fails on new data.

Adjusted 𝑅2 is better for models with multiple predictors since it penalizes unnecessary complexity.

Does Not Assess Prediction Accuracy

A high 𝑅2 doesn’t guarantee accurate predictions, especially in real-world applications.

Sensitive to Outliers

Outliers can artificially inflate or deflate 𝑅2, misleading the interpretation of the model’s strength.

Does Not Measure Individual Variable Importance 𝑅2 tells how much variance is explained collectively, but does not indicate which predictors are most significant.

Cannot Detect Model Misspecification

If the wrong functional form is chosen (e.g., using a linear model for a non-linear relationship), a misleadingly high 𝑅2 might suggest a good fit.


Q20.How can heteroscedasticity be identified in residual plots, and why is it important to address it?

  -> Identifying Heteroscedasticity in Residual Plots

     Residual vs. Fitted Plot: Look for a fan-shaped spread, where residuals start clustered but expand as fitted values increase.

Scale-Location Plot: If residuals show a trend of increasing spread, that’s a strong indicator.

Formal Tests: Statistical tests like the Breusch-Pagan or White test can confirm heteroscedasticity.

Histogram of Residuals: Check if residuals exhibit uneven spread across different value ranges

why It’s Important to Address
Incorrect Standard Errors: Leads to inaccurate hypothesis tests and confidence intervals.

Inefficient Estimates: Even if coefficient estimates remain unbiased, their precision suffers.

Model Misspecification Clue: Could signal missing variables or an incorrect functional form.

Poor Prediction Accuracy: Variance instability distorts future predictions

Q28.What are the limitations of polynomial regression?

  -> Key Limitations
Overfitting Risk

Higher-degree polynomials can fit noise rather than true patterns, leading to unreliable predictions.

The model becomes highly sensitive to small variations in data.

Feature Explosion

As the polynomial degree increases, the number of terms grows exponentially, making computations more complex.

Multivariate polynomial regression can quickly become impractical with too many features.
